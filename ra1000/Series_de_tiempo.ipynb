## Generar Serie de Tiempo ARIMA(1,1,0)

### Subtask:
Generar una serie de tiempo con características ARIMA(1,1,0), que es un proceso de paseo aleatorio con una componente autorregresiva de orden 1 en sus diferencias. Esto implica generar ruido blanco, luego construir una serie AR(1) a partir de este ruido, y finalmente integrar esta serie para obtener la serie ARIMA no estacionaria.


**Reasoning**:
The subtask requires generating an ARIMA(1,1,0) series. This involves defining parameters, generating white noise, simulating an AR(1) process, and then integrating it via cumulative sum to obtain the non-stationary ARIMA series. All these steps can be consolidated into a single code block.



```python
import numpy as np
import pandas as pd

# 1. Define parameters
n = 200  # Number of data points
phi = 0.7  # Autoregressive coefficient for the AR(1) component

# 2. Generate white noise (innovations) Z_t ~ N(0,1)
np.random.seed(42) # for reproducibility
Z_t = np.random.normal(0, 1, n)

# 3. Initialize the AR(1) series
ar_series = np.zeros(n)
ar_series[0] = Z_t[0] # Initial value

# 4. Simulate the AR(1) process from white noise
# ar_series[t] = phi * ar_series[t-1] + Z_t[t]
for t in range(1, n):
    ar_series[t] = phi * ar_series[t-1] + Z_t[t]

# 5. Integrate the ar_series to obtain the ARIMA(1,1,0) series
# ARIMA(1,1,0) implies a single differencing (d=1), so we integrate the AR(1) series.
arima_series = np.cumsum(ar_series)

# 6. Print the first few values of the generated arima_series
print("First 10 values of the simulated ARIMA(1,1,0) series:")
print(arima_series[:10])
```

## Fórmula Matemática del Proceso ARIMA(1,1,0)

Un proceso ARIMA(p,d,q) se define por:

$$(1 - \sum_{i=1}^{p}\phi_i L^i)(1-L)^d X_t = (1 + \sum_{i=1}^{q}\theta_i L^i) \epsilon_t$$

Donde:
*   $X_t$ es la serie de tiempo en el tiempo $t$.
*   $L$ es el operador de retardo (lag operator), tal que $L X_t = X_{t-1}$.
*   $\phi_i$ son los coeficientes autorregresivos.
*   $\theta_i$ son los coeficientes de media móvil.
*   $\epsilon_t$ es el ruido blanco.
*   $p$ es el orden de la parte autorregresiva (AR).
*   $d$ es el orden de diferenciación (I).
*   $q$ es el orden de la parte de media móvil (MA).

Para un proceso **ARIMA(1,1,0)**, tenemos $p=1$, $d=1$, y $q=0$. Esto simplifica la fórmula a:

$$(1 - \phi_1 L)(1-L) X_t = \epsilon_t$$

Expandiendo esto, obtenemos:

$$(1 - L - \phi_1 L + \phi_1 L^2) X_t = \epsilon_t$$

$$X_t - X_{t-1} - \phi_1 X_{t-1} + \phi_1 X_{t-2} = \epsilon_t$$

Reordenando los términos, se puede expresar como:

$$X_t = (1 + \phi_1) X_{t-1} - \phi_1 X_{t-2} + \epsilon_t$$

Alternativamente, si consideramos $Y_t = X_t - X_{t-1}$ como la serie diferenciada (que es un proceso AR(1)), entonces:

$$Y_t = \phi_1 Y_{t-1} + \epsilon_t$$

Y sustituyendo $Y_t = X_t - X_{t-1}$ y $Y_{t-1} = X_{t-1} - X_{t-2}$:

$$X_t - X_{t-1} = \phi_1 (X_{t-1} - X_{t-2}) + \epsilon_t$$

$$X_t = X_{t-1} + \phi_1 (X_{t-1} - X_{t-2}) + \epsilon_t$$

Esta última forma es la que describe cómo se generó la serie: la diferencia actual de $X_t$ es un proceso AR(1).

## Explicar el Modelo Conceptual de Prophet

### Modelo Aditivo Fundamental
Prophet es una librería de Facebook para la predicción de series de tiempo, especialmente diseñada para manejar datos con fuertes componentes estacionales y efectos de días festivos. Su modelo fundamental es un modelo aditivo simple, representado como:

$$y(t) = g(t) + s(t) + h(t) + \epsilon(t)$$

Donde:
*   $y(t)$ es el valor observado de la serie de tiempo en el instante $t$.
*   $g(t)$ es la función de **tendencia** no lineal.
*   $s(t)$ representa los componentes de **estacionalidad** periódicos (diaria, semanal, anual).
*   $h(t)$ es el componente de **días festivos y eventos especiales**.
*   $\epsilon(t)$ es el término de **error**, que representa cualquier variabilidad no explicada por el modelo.

### 1. Componente de Tendencia ($g(t)$)
Prophet modela la tendencia como una función de crecimiento con puntos de cambio (changepoints) que permiten que la tendencia varíe a lo largo del tiempo. Puede utilizar dos modelos de crecimiento:
*   **Crecimiento lineal por tramos (piecewise linear growth):** Es el modelo por defecto y asume que la tendencia cambia abruptamente en ciertos puntos y luego continúa linealmente.
*   **Crecimiento logístico por tramos (piecewise logistic growth):** Utilizado para series de tiempo que muestran un crecimiento saturado (por ejemplo, el crecimiento de usuarios de una plataforma), donde la tendencia se acerca asintóticamente a un límite superior o inferior.

Prophet detecta automáticamente estos puntos de cambio de tendencia, o el usuario puede especificarlos manualmente. La flexibilidad en la modelación de la tendencia es crucial para manejar cambios en el comportamiento a largo plazo de la serie.

### 2. Componente de Estacionalidad ($s(t)$)
La estacionalidad en Prophet se modela utilizando **series de Fourier**. Esto permite capturar patrones repetitivos de diferentes periodicidades, como:
*   **Estacionalidad diaria:** Patrones que se repiten cada día (ej. picos de actividad en ciertas horas).
*   **Estacionalidad semanal:** Patrones que se repiten cada semana (ej. mayor uso de un servicio los fines de semana).
*   **Estacionalidad anual:** Patrones que se repiten cada año (ej. ventas navideñas).

Prophet puede detectar y modelar automáticamente estas periodicidades si están presentes en los datos. La ventaja de las series de Fourier es su capacidad para aproximar cualquier función periódica, lo que hace que Prophet sea muy versátil para diferentes tipos de estacionalidades.

### 3. Componente de Días Festivos ($h(t)$)
Este componente permite incorporar el impacto de días festivos y otros eventos especiales que pueden tener un efecto significativo en la serie de tiempo. Prophet permite:
*   **Días festivos predefinidos:** Para muchos países, Prophet tiene listas de días festivos predefinidas que se pueden agregar fácilmente al modelo.
*   **Eventos personalizados:** Los usuarios pueden definir sus propios eventos o fechas importantes (por ejemplo, lanzamientos de productos, campañas publicitarias) y Prophet modelará su impacto aditivo en la serie de tiempo. Cada día festivo se trata como un efecto puntual con una duración específica.

### 4. Término de Error ($\epsilon(t)$)
El término de error representa la varianza residual que no puede ser explicada por los otros componentes del modelo. Se asume que estos errores son ruido blanco, es decir, son variables aleatorias independientes e idénticamente distribuidas con media cero.

### Ventajas de Prophet
Prophet es popular por varias razones:
*   **Robusto a datos faltantes y atípicos:** Puede manejar series de tiempo con huecos en los datos o valores extremos sin necesidad de imputación previa.
*   **Manejo automático de cambios de tendencia:** Su enfoque de crecimiento por tramos detecta y se adapta a cambios significativos en la tendencia de la serie.
*   **Múltiples estacionalidades:** Permite modelar simultáneamente patrones diarios, semanales, anuales, etc., de forma flexible.
*   **Incorporación de días festivos y eventos:** Facilita la inclusión de influencias externas conocidas que impactan la serie.
*   **Fácil de usar:** Su API está diseñada para ser intuitiva y accesible incluso para usuarios sin un conocimiento profundo de series de tiempo.

## Analizar ACF y PACF de la Serie Original

### Subtask:
Calcular y graficar las funciones de autocorrelación (ACF) y autocorrelación parcial (PACF) de la serie de precios de cierre original de Amazon para entender su estructura de dependencia.


**Reasoning**:
The subtask requires calculating and plotting the ACF and PACF of the `precios_amzn` series. I need to import the necessary functions from `statsmodels.graphics.tsaplots` and then create the plots with appropriate titles and labels.



```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt

# Plot ACF
plt.figure(figsize=(12, 6))
plot_acf(precios_amzn, lags=40, ax=plt.gca())
plt.title('Función de Autocorrelación (ACF) para Precios de Cierre de AMZN')
plt.xlabel('Lag')
plt.ylabel('Autocorrelación')
plt.grid(True)
plt.show()

# Plot PACF
plt.figure(figsize=(12, 6))
plot_pacf(precios_amzn, lags=40, ax=plt.gca())
plt.title('Función de Autocorrelación Parcial (PACF) para Precios de Cierre de AMZN')
plt.xlabel('Lag')
plt.ylabel('Autocorrelación Parcial')
plt.grid(True)
plt.show()
```

**Reasoning**:
The previous code failed because the number of specified lags (40) was greater than the number of data points in `precios_amzn` (22), leading to a `ValueError`. To fix this, I will adjust the `lags` parameter to be a value appropriate for the length of the series, specifically `min(len(precios_amzn) // 2 - 1, 10)` which will set it to a maximum of 10 for this series, ensuring valid plotting.



```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt

# Determine an appropriate number of lags based on the series length
# A common rule of thumb is lags < N/2, and generally not too many if N is small.
max_lags = min(len(precios_amzn) // 2 - 1, 10) # Set to max 10 lags or N/2 - 1 if N is smaller

# Plot ACF
plt.figure(figsize=(12, 6))
plot_acf(precios_amzn, lags=max_lags, ax=plt.gca())
plt.title('Función de Autocorrelación (ACF) para Precios de Cierre de AMZN')
plt.xlabel('Lag')
plt.ylabel('Autocorrelación')
plt.grid(True)
plt.show()

# Plot PACF
plt.figure(figsize=(12, 6))
plot_pacf(precios_amzn, lags=max_lags, ax=plt.gca())
plt.title('Función de Autocorrelación Parcial (PACF) para Precios de Cierre de AMZN')
plt.xlabel('Lag')
plt.ylabel('Autocorrelación Parcial')
plt.grid(True)
plt.show()
```
